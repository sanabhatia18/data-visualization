```python
import plotly.express as px
```

```python
import seaborn as sns
flight=sns.load_dataset('flights')
flight
```

```python
monthly_pass=flight.groupby('month').agg({'passengers':'sum'}).reset_index()
monthly_pass
```

```python
import plotly.express as px
px.line(monthly_pass,x='month',y='passengers')
```

```python
tip=px.data.tips()
tip
```

```python
px.bar(tip,x='sex',y='total_bill',color='smoker')
```

```python
px.bar(tip,x='sex',y='total_bill',color='smoker',barmode='group')
#barmode bbars different krdeta
```
