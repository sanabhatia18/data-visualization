```python
#new data
month = ['jan','feb','march','apr','may','june','july','aug','sept','oct','nov','dec']
sales = [1111,2222,3333,4444,5555,6666,7777,8888,9999,1000,1122,1133]
profit= [100,200,300,400,500,600,700,800,900,111,112,113]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(15,7))  #canvas size
plt.plot(month,sales,linestyle='dotted',color='green',label='SALES')
plt.fill_between(month,profit,linestyle='solid',color='blue',label='PROFIT')
plt.title('monthlysales')
plt.ylabel('sales')
plt.xlabel('month')
plt.gca().set_facecolor('gray')   #background color
plt.legend()
plt.show()
```
