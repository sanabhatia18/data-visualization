```python
import matplotlib.pyplot as plt
```

```python
regions=['north','east','west','south']
sales=[1227,3441,6553,8872]
```

```python
plt.pie(x=sales,labels=regions)
```

```python
plt.pie(x=sales,labels=regions,autopct='%1.1f%%')
plt.show()
```

```python
month = ['jan','feb','march','apr','may','june','july','aug','sept','oct','nov','dec']
sales = [1111,2222,3333,4444,5555,6666,7777,8888,9999,1000,1122,1133]
profit= [100,200,300,400,500,600,700,800,900,111,112,113]
```

```python
pie_regions=['north','east','west','south']
pie_sales=[1227,3441,6553,8872]
pie_category=['tech','officesupp','paper','automob']
pie_sale=[9494,2323,6373,1839]
```

```python
bar_products=['tshits','cap','belts','cargo','chinos','trouser','shoes']
bar_sales=[3344,2444,6788,4555,9986,4567,8765]
```

```python
# plt.figure(figsize=(20,4)) ese ni likhna
fix,x = plt.subplots(2,2,figsize=(15,15))
x[0,0].plot(month,sales)
x[0,1].bar(bar_products,bar_sales)
x[1,0].pie(labels=pie_regions,x=pie_sales,autopct='%1.0f%%')  #1.0 pe round ayega
x[1,1].pie(labels=pie_category,x=pie_sale,autopct='%1.1f%%')  #1.1 pe decimal ayega
plt.show()
```
