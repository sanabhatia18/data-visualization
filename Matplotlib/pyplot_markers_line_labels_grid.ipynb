```python
#library import

import matplotlib.pyplot as plt  
```

```python
#data
month = ['jan','feb','march','apr','may','june','july','aug','sept','oct','nov','dec']
sales = [111,222,333,444,555,666,777,888,999,100,112,113]
```

```python
len(month)
```

```python
len(sales)
```

```python
plt.plot(month,sales)
```

```python
plt.plot(month,sales,linestyle='dashed') 
```

```python
plt.plot(month,sales,linestyle='dotted')
```

```python
plt.plot(month,sales,linestyle='dotted',color='pink')
```

```python
plt.plot(month,sales,linestyle='dotted',color='green',label='SALES')
plt.legend()
```

```python
plt.figure(figsize=(15,10))
plt.plot(month,sales,linestyle='dotted',color='purple',label='SALES')
plt.title('monthly sales')
plt.ylabel('sales')
plt.xlabel('month')
plt.legend()
plt.show()
```
