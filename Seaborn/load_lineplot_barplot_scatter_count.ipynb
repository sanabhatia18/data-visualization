```python
fl= sns.load_dataset('flights')
iris = sns.load_dataset('iris')
tips= sns.load_dataset('tips')
print(fl)
```

```python
sns.lineplot(fl,x='month',y='passengers',linestyle='dotted',errorbar=("ci", 20))

#95 errorbar is default ye ci hota -confidence interval jo blue gap hai
```

```python
ax=sns.barplot(tips,x='time',y='total_bill')
for i in ax.containers:
    ax.bar_label(i)
```

```python
ax=sns.barplot(tips,x='time',y='total_bill',errorbar=('ci',0))
for i in ax.containers:
    ax.bar_label(i)

#here errorbar removes the black line
```

```python
ax=sns.countplot(x=tips['sex'])
for i in ax.containers:
    ax.bar_label(i)
```

```python
ax=sns.scatterplot(tips,
                   x='total_bill',
                   y='tip',
                   hue='sex',
                   palette=['blue','pink'])

#hue jis ke acc color chahiye
#palette jo color chahiye
```
