HEATMAP

```python
import seaborn as sns
```

```python
iris=sns.load_dataset('iris')
iris
```

```python
iris.select_dtypes(float)  #float type shows
```

```python
correlation=iris.select_dtypes(float).corr()  #correlation shows
correlation
```

```python
sns.heatmap(correlation,annot=True)  #labels dikhenge
```
